# Seaborn Library exploration

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "Days": [1,2,3,4,5],
    "NOP": [50,40,60,30,44]
}
df = pd.DataFrame(data)
df
```

```python
sns.lineplot(data=data,x="Days",y="NOP")
```

```python
data = {
    "Days": [1,2,3,4,5],
    "NOP": [50,40,60,30,44]
}
df = pd.DataFrame(data)
print(df)

sns.barplot(data=data,x='Days',y='NOP',hue='Days',palette="spring",order=[5,4,3,2,1],errorbar = ("ci",0))
plt.show()
```

```python
data = sns.load_dataset('tips')
data
```

```python
sns.scatterplot(data=data,x='total_bill',y='tip',hue = 'day',size='size',marker = 'o',palette='viridis')
plt.show()
```

```python
data = sns.load_dataset('tips')
gp = data.groupby("day").agg({'tip':'mean'})
sns.heatmap(gp)
plt.show()
```

```python
data = sns.load_dataset('tips')
sns.countplot(data = data,x = 'day',hue = 'day')
```

```python
sns.violinplot(data = data , x = 'day',y = 'tip')
plt.show()
```

```python
sns.boxplot(data=data,x='tip',y='sex',orient='vertical')
plt.show()
```

```python
sns.pairplot(data,hue='day')
plt.show()
```

```python
sns.stripplot(data = data,x = 'day',y = 'tip')
```

```python
data = sns.load_dataset('exercise')
sns.barplot(x = 'time',y = 'pulse',data=data,hue='time')
plt.show()
```

```python
sns.palplot(sns.color_palette())
plt.show()
```

```python
sns.palplot(sns.color_palette('viridis',5))
```

```python
data = sns.load_dataset('tips')
a = sns.FacetGrid(data,col = 'smoker',height = 3,hue = 'sex')
a.map(sns.barplot,'day','tip')
plt.show()
```

```python
sns.relplot(data = data,x = 'tip',y = 'total_bill',hue = 'sex',col = 'day')
plt.show()
```

```python
sns.swarmplot(data = data,x = 'total_bill',y = 'day',hue = 'sex')
plt.show()
```

```python
sns.stripplot(data = data,x = 'day',y = 'total_bill',jitter = 0.3,hue = 'sex')
plt.show()
```
